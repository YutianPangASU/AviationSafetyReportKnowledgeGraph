```python
import pandas as pd
pd.set_option('display.max_colwidth', None)
import pandas_access as mdb

db_filename = './NTSBData/avall.mdb'

# Listing the tables.
print('Showing list of tables in the NSTB MS Access database:')
for tbl in mdb.list_tables(db_filename):
  print(tbl)

# Read the narratives table only.
df = mdb.read_table(db_filename, "narratives")
```

```python
df = df.dropna(axis=0, how='any', subset=['narr_accp', 'narr_accf', 'narr_cause']).drop_duplicates()
df['Text'] = df[['narr_accp', 'narr_accf', 'narr_cause']].agg('  '.join, axis=1)
df = df.drop(columns=['narr_accp', 'narr_accf', 'narr_cause', 'Aircraft_Key', 'narr_inc', 'lchg_userid', 'lchg_date']).replace('\n', '', regex=True).replace('\r', '', regex=True)
df['ac_id'] = df.Text.str.extract(r', ([N][\w]+), ', expand=True)
df.insert(0, 'ac_id', df.pop('ac_id'))
```
